```python
import requests
from bs4 import BeautifulSoup
import re
from youtube_transcript_api import YouTubeTranscriptApi
import os
```

```python
def save_transcript_to_file(video_url, languages_list, file_save_as):

    # --------------------------------------------------
    # 1. Extract Video ID
    # --------------------------------------------------
    def get_video_id(url):
        match = re.search(
            r"(?:v=|youtu\.be/|shorts/)([^&?/]+)",
            url
        )
        if match:
            return match.group(1)
        raise ValueError("Invalid YouTube URL")

    video_id = get_video_id(video_url)

    # --------------------------------------------------
    # 2. Extract Video Title
    # --------------------------------------------------
    def get_video_title(url):
        response = requests.get(
            "https://www.youtube.com/oembed",
            params={"url": url, "format": "json"},
            timeout=30
        )
        response.raise_for_status()
        return response.json()["title"]

    video_title = get_video_title(video_url)

    # --------------------------------------------------
    # 3. Download Transcript
    # --------------------------------------------------
    ytt_api = YouTubeTranscriptApi()

    transcript_data = ytt_api.fetch(
        video_id,
        languages=languages_list
    )

    # Extract text from transcript snippets
    transcricao = " ".join(
        snippet.text for snippet in transcript_data
    )

    # --------------------------------------------------
    # 4. Create Video Information
    # --------------------------------------------------
    infos_video = f"""
    Video Information

    Title: {video_title}
    URL: {video_url}

    Transcript:
    {transcricao}
    """

    # --------------------------------------------------
    # 5. Save Transcript to Text File
    # --------------------------------------------------
    os.makedirs(os.path.dirname(file_save_as) or ".", exist_ok=True)

    with open(file_save_as, "w", encoding="utf-8") as f:
        f.write(infos_video)

    print(f"Transcript successfully saved to {file_save_as}")

    return infos_video
```

```python
video_url = "https://www.youtube.com/watch?v=II28i__Tf3M"
languages_list = ["pt"]
file_save_as = "data_store/Introduction to BFF.txt"

save_transcript_to_file(video_url, languages_list, file_save_as)
```
